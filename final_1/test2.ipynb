BA无标度网络 MMCA方法和MC方法对比

```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import warnings
# 忽略FutureWarning
warnings.filterwarnings("ignore", category=FutureWarning)

MC_p0_3 = pd.read_csv("./test2_data/BA_MC_p_0.3.csv")
MC_p0_6 = pd.read_csv("./test2_data/BA_MC_p_0.6.csv")
MC_p0_9 = pd.read_csv("./test2_data/BA_MC_p_0.9.csv")

MMCA_p03 = pd.read_csv("./test2_data/BA_MMCA_p_0.3.csv")
MMCA_p06 = pd.read_csv("./test2_data/BA_MMCA_p_0.6.csv")
MMCA_p09 = pd.read_csv("./test2_data/BA_MMCA_p_0.9.csv")

```

```python
plt.figure(figsize=(9, 6))

sns.lineplot(data = MMCA_p03, x = 'beta', linewidth=2, markersize=5, color='#56cfe1',
             y= 'final_infected_proportion' ,label=r'0.3 (Model)')
sns.lineplot(data = MMCA_p06, x = 'beta', linewidth=2, markersize=5, color='#1565c0',
             y= 'final_infected_proportion' ,label=r'0.6 (Model)')
sns.lineplot(data = MMCA_p09, x = 'beta', linewidth=2, markersize=5, color='#d81159',
             y= 'final_infected_proportion' ,label=r'0.9 (Model)')

sns.scatterplot(data = MC_p0_3, x = 'beta', s = 80, color='#56cfe1',
             y= 'final_infected_proportion' ,label=r'0.3 (MC)',marker='o')
sns.scatterplot(data = MC_p0_6, x = 'beta', s = 80, color='#1565c0',
             y= 'final_infected_proportion' ,label=r'0.6 (MC)',marker='p')
sns.scatterplot(data = MC_p0_9, x = 'beta', s = 80, color='#d81159',
             y= 'final_infected_proportion' ,label=r'0.9 (MC)',marker='d')

plt.axvline(x=0.0021, color='#56cfe1', linestyle='--', linewidth=1)
plt.axvline(x=0.0019, color='#1565c0', linestyle='--', linewidth=1)
plt.axvline(x=0.0017, color='#d81159', linestyle='--', linewidth=1)

plt.xlabel(r'$\ \beta$',fontsize='30')
plt.ylabel(r'$\ \rho$',fontsize='30')
plt.xscale("log")
plt.yscale("linear")
plt.ylim(-0.05, 0.95)  # 设置y轴范围
plt.yticks(np.arange(0.0, 0.95, 0.2), fontsize='20')
plt.xticks([0.001, 0.01, 0.1], [ "$10^{-3}$", "$10^{-2}$", "$10^{-1}$"], fontsize='20')
plt.xlim(0.0002, 0.7)  # 设置x轴范围
# plt.title('Final Infected Proportion vs. Beta for different p_move values')
# 添加图例，并设置图例标题
legend = plt.legend(title='Mobility Probability (p)', loc='lower right',fontsize='18')

# 设置图例标题字体大小
plt.setp(legend.get_title(), fontsize='18')

# 设置图例框边框
legend.get_frame().set_edgecolor('black')
legend.get_frame().set_linewidth(1)
plt.savefig('BA_MMCA_MC.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
