```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import matplotlib.ticker as ticker

# 忽略FutureWarning
warnings.filterwarnings("ignore", category=FutureWarning)

# 读取CSV文件
all_results_df = pd.read_csv('./data2/all_results_matrix_1007.csv', index_col=0)

# 获取p_move_values和alpha_values
p_move_values = all_results_df.columns.astype(float)
alpha_values = all_results_df.index.astype(float)

# 绘制图表，分别为每个alpha值绘制曲线
plt.figure(figsize=(9, 6))

# 绘制alpha=0.4的曲线
sns.lineplot(x=p_move_values, y=all_results_df.loc[0.4]*1000, linewidth=2, markersize=9, color='#56cfe1', marker='p',  
               label=r'$\alpha$=0.4') 
sns.lineplot(x=p_move_values, y=all_results_df.loc[0.5]*1000, linewidth=2, markersize=12, color='#1565c0', marker='*',
             label=r'$\alpha$=0.5')
sns.lineplot(x=p_move_values, y=all_results_df.loc[0.6]*1000, linewidth=2, markersize=9, color='#d81159', marker ='v',
             label=r'$\alpha$=0.6')


# 添加图表标签
plt.xlabel(r'$p$',fontsize='30')
plt.ylabel(r'$\beta_c$',fontsize='30')
# plt.title('Critical Beta vs. p_move for different alpha values')

# 设置y轴范围
plt.ylim(1.4, 2.8)

plt.xscale("log")
plt.xticks([0.01, 0.1, 1], ["$10^{-2}$", "$10^{-1}$", "$10^{0}$"], fontsize='20')
plt.yticks(fontsize='20')

# # 自定义 y 轴格式化为数值部分
# def format_y_ticks(x, pos):
#     return f'{x:.1f}'  # 保留一位小数

# # 设置y轴为自定义格式
# plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(format_y_ticks))



# 在 y 轴顶部添加统一的指数部分
plt.gca().text(-0.0, 1.0, '1e-03', fontsize='18', verticalalignment='bottom', transform=plt.gca().transAxes)
plt.axvline(x=0.6, color='#a4133c', linestyle='--', linewidth=1)
# 显示图例
plt.legend(fontsize='16')
plt.savefig('final_3b.pdf', dpi=300, bbox_inches='tight')
# 显示图表
plt.show()
```
